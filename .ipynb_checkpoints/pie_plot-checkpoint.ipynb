```python
import plotly.graph_objects as go
```

```python
labels = ["Em Execução", "Aguardando Material", "Aguardando Atendimento", "Em Avaliação"]
values = [20, 20, 37, 27]
```

```python
fig = go.Figure(data = go.Pie(values=values, labels=labels, pull=[0,0,0.1,0]))

fig.update_traces(hoverinfo="label+percent", textinfo="value+percent",textfont_size=12)
fig.show()
```
