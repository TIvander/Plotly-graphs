```python
import plotly.graph_objects as go
import numpy as np
```

```python
t = np.linspace(0, 10, 100)
y= np.sin(t)
```

```python
fig = go.Figure(data = go.Scatter(x=t, y=y, mode="markers"))
fig.show()
```

```python
n = 100
x = np.linspace(0, 1, n)
random_y0 = np.random.randn(len(x)) + 5
random_y1 = np.random.randn(len(x)) 
random_y2 = np.random.randn(len(x)) - 5
```

```python
fig = go.Figure()

fig.add_traces(go.Scatter(x=x, y=random_y0, mode="markers", name="marker"))

fig.add_traces(go.Scatter(x=x, y=random_y1, mode="markers+lines", name="markers + lines"))

fig.add_traces(go.Scatter(x=x, y=random_y2,mode="lines", name="lines"))

fig.show()
```

```python
fig = go.Figure(data=go.Scatter(
    x=[1,2,3,4],
    y=[10,11,12,13],
    mode='markers',
    marker=dict(size=[40, 60, 80, 100]),
    hovertemplate="Valor total R$ %{y} - %{marker.size}"    
))
fig.show()
```
